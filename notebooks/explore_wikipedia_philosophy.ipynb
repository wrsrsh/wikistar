# Explore the Wikipedia → Philosophy graph

This notebook queries the persistent outputs created by `wikipedia_philosophy_graph.py`; it does **not** download or parse Wikipedia again. Before opening it, deploy the app from the project directory:

```bash
modal deploy wikipedia_philosophy_graph.py
modal launch jupyter --gpu a10g
```

The Jupyter GPU is optional for these query cells. The pipeline itself uses a T4 Modal GPU to generate its force-directed SVG overview.

````python
%pip install -q modal pandas

import modal
import pandas as pd

RUN_ID = "enwiki-latest-2026-08"  # Set this to the completed pipeline run.
APP_NAME = "wikipedia-philosophy-graph"

sample_routes = modal.Function.from_name(APP_NAME, "inspect_routes")
route_for = modal.Function.from_name(APP_NAME, "lookup_route")
````

````python
# Inspect a bounded neighborhood nearest Philosophy.
rows = sample_routes.remote(RUN_ID, limit=100, max_distance=5)
pd.DataFrame(rows)
````

````python
# Ask for the exact deterministic path for an article.
result = route_for.remote(RUN_ID, "Artificial intelligence")
result
````
